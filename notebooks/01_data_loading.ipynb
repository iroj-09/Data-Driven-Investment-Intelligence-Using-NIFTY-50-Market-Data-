```python
metadata = pd.read_csv("../data/metadata/stock_metadata.csv")

metadata.head()
```

```python
symbol_mapping = {
    "MUNDRAPORT": "ADANIPORTS",
    "UTIBANK": "AXISBANK",
    "BAJAUTOFIN": "BAJFINANCE",
    "BHARTI": "BHARTIARTL",
    "HEROHONDA": "HEROMOTOCO",
    "HINDALC0": "HINDALCO",
    "HINDLEVER": "HINDUNILVR",
    "INFOSYSTCH": "INFY",
    "JSWSTL": "JSWSTEEL",
    "KOTAKMAH": "KOTAKBANK",
    "TELCO": "TATAMOTORS",
    "TISCO": "TATASTEEL",
    "UNIPHOS": "UPL",
    "SESAGOA": "VEDL",
    "SSLT": "VEDL",
    "ZEETELE": "ZEEL"
}
```

```python
stocks_df["Symbol"] = stocks_df["Symbol"].replace(symbol_mapping)
```

```python
stock_symbols = set(stocks_df["Symbol"].unique())
metadata_symbols = set(metadata["Symbol"].unique())

print("Symbols missing in metadata:")
print(stock_symbols - metadata_symbols)
```

```python
stocks_df = stocks_df.merge(
    metadata,
    on="Symbol",
    how="left"
)
```

```python
(stocks_df["Series_x"] == stocks_df["Series_y"]).value_counts(dropna=False)
```

```python
# Rename Series_x to Series
stocks_df.rename(columns={"Series_x": "Series"}, inplace=True)

# Drop the duplicate column
stocks_df.drop(columns=["Series_y"], inplace=True)
```

```python
stocks_df.columns
```

```python
stocks_df = stocks_df.sort_values(
    by=["Symbol", "Date"]
).reset_index(drop=True) 
```
